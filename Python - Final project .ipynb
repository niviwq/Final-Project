```python
cap_orders["order_date"] = pd.to_datetime(cap_orders["order_date"], format="%d-%m-%Y")
```

```python
analyse_date = cap_orders["order_date"].max() + pd.Timedelta(days=1)
analyse_date
```

```python
rfm=pd.merge(cap_orders,cap_order_items,on="order_id",how="right")
rfm
```

```python
rfm = rfm.groupby("customer_id").agg(
    Recency = ("order_date", lambda x: (analyse_date - x.max()).days),
    Frequency = ("order_id", "nunique"),
    Monetary = ("total_price", "sum")
).reset_index()
```

```python
rfm
```

```python
rfm ["R_Score"] = pd.qcut(rfm["Recency"], 4, labels= [4,3,2,1])
rfm ["F_Score"] = pd.qcut(rfm["Frequency"].rank(method = "first"),4,labels=[1,2,3,4])
rfm ["M_Score"] = pd.qcut(rfm["Monetary"].rank(method = "first"),4,labels=[1,2,3,4])
```

```python
rfm["RFM_Score"] = (
    rfm["R_Score"].astype(int)
    + rfm["F_Score"].astype(int)
    + rfm["M_Score"].astype(int))
```

```python
rfm
```

```python
def segment_customer(score):
    if score>=10:
        return"Champions"
    elif score>= 8:
        return "Loyal Customers"
    elif score>= 6 :
        return "Potential Loyalist"
    elif score<=4:
        return "At Risk"
    else:
        return "Lost Customers"
rfm["Segment"] = rfm ["RFM_Score"].apply(segment_customer)
```

```python
rfm
```

```python
customer_segment = rfm[["customer_id","Recency","Frequency","Monetary","Segment"]]
```

```python
customer_segment
```

```python
customer_segment.to_sql(
    "customer_segment",
    con=sql_link,
    if_exists="replace",
    index=False)
```
